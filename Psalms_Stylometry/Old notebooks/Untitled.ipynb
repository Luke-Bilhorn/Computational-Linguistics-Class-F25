```python
import pandas as pd
from rpy2 import robjects as ro
from rpy2.robjects import StrVector, ListVector
from rpy2.robjects.packages import importr

# Import stylo
stylo = importr("stylo")

# Minimal fake dataframe
chapters_df = pd.DataFrame({
    "chapter_name": ["chapter1", "chapter2"],
    "full_chapter_text": [
        "This is the text of chapter one.",
        "This is the text of chapter two."
    ]
})

# Convert to R list with names and StrVectors
texts = chapters_df["full_chapter_text"].tolist()
names = chapters_df["chapter_name"].tolist()
r_list = ListVector({names[i]: StrVector([texts[i]]) for i in range(len(texts))})

# Send to R environment
ro.globalenv["direct_corpus"] = r_list

# Run stylo using the direct corpus
stylo.stylo(
    corpus_type="direct",
    corpus=ro.globalenv["direct_corpus"]
)
```
