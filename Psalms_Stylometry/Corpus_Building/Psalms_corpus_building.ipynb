```python
import re
import pandas as pd

def parse_line(line):
    """
    Parse a single raw verse line of the form:
    ' 1  ׃1   אַשְׁרֵי...'
    Returns: chapter, verse, text
    """
    line = line.strip()
    
    # Match: verse_number, chapter_number, verse_text
    # Example line: "1  ׃1   אַשְׁרֵי ..."
    m = re.match(r"(\d+)\s+׃(\d+)\s+(.*)", line)
    if not m:
        return None  # skip malformed
    
    verse = int(m.group(1))
    chapter = int(m.group(2))
    text = m.group(3).strip()
    return chapter, verse, text


def extract_author(text):
    """
    Very simple author guess:
    If first verse contains 'לְדָוִד' or similar phrases.
    This can be expanded as needed.
    """
    if "לְדָוִד" in text or "לדוד" in text:
        return "David"
    if "לְשְׁלֹמֹה" in text:
        return "Solomon"
    return "Anonymous"


# ---- Read and parse the raw text file ----

rows = []
with open("Psalms.txt", "r", encoding="utf-8") as f:
    #next(f)  # skip the first line to get rid of any authors mentioned.
    for line in f:
        parsed = parse_line(line)
        if parsed:
            rows.append(parsed)

df = pd.DataFrame(rows, columns=["chapter", "verse", "text"])


# ---- Determine author per chapter ----
#authors = {}

#for chapter, first_verse_text in df.sort_values("verse").groupby("chapter").first()["text"].items():
    #authors[chapter] = extract_author(first_verse_text)

#for chapter, first_verse_text in df.sort_values("verse").groupby("chapter").first()["text"].items():
#    print(first_verse_text)

#df["author"] = df["chapter"].map(authors)

# ---- Assign authors from the manual mapping ----
df["author"] = df["chapter"].map(manual_authors).fillna("Anonymous")




# ---- Combine verses into a single string per chapter ----

chapters_df = (
    df.groupby(["chapter", "author"])["text"]
      .apply(lambda verses: " ".join(verses))
      .reset_index(name="full_chapter_text")
)
print(chapters_df)
```

```python
import os

# Ensure authors are correctly mapped per chapter
df["author"] = df["chapter"].map(manual_authors).fillna("Anonymous")

# Combine all verses per chapter
chapters_df = (
    df.groupby("chapter")
      .agg({
          "author": "first",  # assumes one author per chapter
          "text": lambda verses: " ".join(verses)  # combine all verses
      })
      .reset_index()
      .rename(columns={"text": "full_chapter_text"})
)

# Create directory
output_dir = "stylo_corpus"
os.makedirs(output_dir, exist_ok=True)

# Write each chapter as a separate file
for _, row in chapters_df.iterrows():
    chapter_num = row["chapter"]
    author = row["author"].replace(" ", "_")
    text = row["full_chapter_text"]
    
    filename = f"{chapter_num:03d}_{author}.txt"
    filepath = os.path.join(output_dir, filename)
    
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(text)

print(f"Corpus written to {output_dir} with {len(chapters_df)} files.")
```
